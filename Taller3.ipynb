```python
limite = 1000
cuenta = 0
vueltas = 0

for n in range(2, limite + 1):
    es_primo = True
    d = 2
    while d * d <= n and es_primo:
        vueltas += 1
        if n % d == 0:
            es_primo = False
        d += 1
    if es_primo:
        cuenta += 1

print(cuenta)
print(vueltas)
```

```python
n = 10

# Traducción A
a = 1
b = 1
for i in range(3, n):
    aux = a + b
    a = b
    b = aux
print("A:", b)

# Traducción B
a = 1
b = 1
for i in range(3, n + 1):
    a, b = b, a + b
print("B:", b)
```
